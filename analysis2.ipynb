```python
t = 0
img = np.asarray(vol[t]).max(0)
pts = nodes[nodes["t"] == t]

plt.imshow(img, cmap="gray")
plt.scatter(pts["x"], pts["y"], c="cyan", s=20)
plt.show()
```

```python
import cv2

t = 0
gray0 = cv2.normalize(np.asarray(vol[t]).max(0), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
gray1 = cv2.normalize(np.asarray(vol[t+1]).max(0), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

pts0 = nodes[nodes["t"] == t][["x", "y"]].to_numpy().astype(np.float32)
pts_gt1 = nodes[nodes["t"] == t+1][["x", "y"]].to_numpy().astype(np.float32)
```

```python
p0 = pts0.reshape(-1, 1, 2)
p1, st, _ = cv2.calcOpticalFlowPyrLK(gray0, gray1, p0, None)
pred = p1[st.ravel() == 1].reshape(-1, 2)
print(pred)
```

```python
plt.imshow(gray1, cmap="gray")
plt.scatter(pts_gt1[:, 0], pts_gt1[:, 1], c="cyan", s=30, label="GT t+1")
plt.scatter(pred[:, 0], pred[:, 1], c="red", s=20, marker="x", label="OF pred")
plt.legend()
plt.show()
```

```python
plt.imshow(gray1, cmap="gray")
plt.scatter(pts_gt1[:, 0], pts_gt1[:, 1], c="cyan", s=30, label="GT t+1")
plt.scatter(pred[:, 0], pred[:, 1], c="red", s=20, marker="x", label="OF pred")

for (x0, y0), (x1, y1) in zip(pts0, pred):
    plt.arrow(x0, y0, x1 - x0, y1 - y0, color="yellow", head_width=3, length_includes_head=True)

plt.legend()
plt.show()
```
