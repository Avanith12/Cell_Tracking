```python
import numpy as np
import pandas as pd
import zarr
import matplotlib.pyplot as plt
```

# 44b6_0113de3b



```python
vol = zarr.open(
    "/hpcstor6/scratch01/a/a.kanamarlapudi001/cell_tracking/biohub-cell-tracking-during-development/test/44b6_0113de3b.zarr",
    mode="r",
)["0"]
```

```python
vol.shape
```

```python
T = vol.shape[0]
all_t = list(range(T))
print("number of times T =", T)
print("all t values =", all_t)
```

```python
t = 50
img = np.asarray(vol[t]).max(0)

plt.imshow(img, cmap="gray")
plt.title("44b6_0113de3b MIP")
plt.show()
```

```python
g = zarr.open(
    "/hpcstor6/scratch01/a/a.kanamarlapudi001/cell_tracking/biohub-cell-tracking-during-development/train/44b6_0113de3b.geff",
    mode="r",
)
```

```python
nodes = pd.DataFrame({
    "id": g["nodes/ids"][:],
    "t": g["nodes/props/t/values"][:],
    "z": g["nodes/props/z/values"][:],
    "y": g["nodes/props/y/values"][:],
    "x": g["nodes/props/x/values"][:],
})

labeled_t = sorted(nodes["t"].unique().tolist())
print("t values that HAVE GT labels =", labeled_t)
print("count of labeled times =", len(labeled_t))

nodes
```

```python
pts = nodes[nodes["t"] == t]
pts
```

```python
plt.imshow(img, cmap="gray")
plt.scatter(pts["x"], pts["y"], c="cyan", s=20)
plt.title("overlay GT (from train geff)")
plt.show()
```
